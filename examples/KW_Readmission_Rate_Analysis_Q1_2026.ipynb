# Hospital Readmission Rate Analysis — Q1 2026 vs Q4 2025

KW, a data scientist at Singapore's Ministry of Health, is tasked with analysing why the readmission rate across all hospitals in Singapore has increased significantly in Q1 2026 compared to Q4 2025, and recommending what MOH should do next.

KW applies his 4-tier analytics framework:
- **Descriptive** — what does the data show?
- **Diagnostic** — why did it happen?
- **Predictive** — where is it heading?
- **Prescriptive** — what should MOH do?

Statistical significance is determined using proportion testing via the margin of error formula:
> MoE = z × √(p × (1 − p) / n)

```python
import json
import sys
sys.path.insert(0, '..')

import tinytroupe
from tinytroupe.agent import TinyPerson
from tinytroupe.environment import TinyWorld
from tinytroupe.extraction import ResultsExtractor
```

## Load KW and apply his analytics fragment

```python
kw = TinyPerson.load_specification("./agents/KW.agent.json")
kw + "./fragments/moh_health_sentiment_analyst.fragment.json"

kw.minibio()
```

## Set up the scenario

We provide KW with the raw data briefing — discharge counts and readmission rates for both quarters, along with contextual signals from hospital operations, seasonal factors, and policy changes.

```python
world = TinyWorld("MOH Analytics Room", [kw])
world.make_everyone_accessible()

world.broadcast_internal_goal(
    "You are in an analytical deep-dive session at MOH. Senior leadership has flagged "
    "a concerning rise in hospital readmission rates in Q1 2026. You have been provided "
    "with the data and contextual briefing below. Your job is to conduct a rigorous "
    "4-tier analysis — descriptive, diagnostic, predictive, and prescriptive — and "
    "present your findings clearly. You must confirm statistical significance using "
    "proportion testing and the margin of error formula before drawing any conclusions."
)
```

## Step 1 — Descriptive Analysis

Provide KW with the raw numbers and ask him to establish what the data shows, including significance testing.

```python
descriptive_briefing = """
KW, here is the readmission data for your review:

HOSPITAL READMISSION DATA — SINGAPORE
--------------------------------------
Q4 2025:
  - Total hospital discharges: 44,200
  - Readmissions within 30 days: 3,713
  - Readmission rate: 8.4%

Q1 2026:
  - Total hospital discharges: 42,800
  - Readmissions within 30 days: 4,408
  - Readmission rate: 10.3%

Breakdown by hospital cluster (Q1 2026 vs Q4 2025):
  - National University Hospital cluster:  10.8% vs 8.6%
  - Singapore General Hospital cluster:    10.5% vs 8.7%
  - Tan Tock Seng Hospital cluster:         9.9% vs 8.2%
  - Changi General Hospital:                9.8% vs 8.1%
  - Ng Teng Fong General Hospital:         10.1% vs 8.3%

Breakdown by top readmission diagnoses (Q1 2026):
  - Heart failure:              22% of readmissions
  - Pneumonia / respiratory:    18% of readmissions
  - Hip fracture / orthopaedic: 14% of readmissions
  - Diabetes complications:     12% of readmissions
  - Other:                      34% of readmissions

Please confirm whether the increase is statistically significant using proportion testing 
and the margin of error formula. Then describe what the data is telling us.
"""

kw.listen_and_act(descriptive_briefing, max_content_length=4000)
```

## Step 2 — Diagnostic Analysis

Provide KW with contextual signals and ask him to identify the root causes.

```python
diagnostic_briefing = """
Good. Now here are the contextual signals we have gathered from hospital operations, 
seasonal data, and policy records for Q1 2026:

SEASONAL & POPULATION FACTORS:
  - Chinese New Year fell in Q1 2026 (late January). Historically, post-CNY periods 
    see a surge in elderly admissions due to disrupted routines, medication non-adherence 
    during festive periods, and delayed care-seeking.
  - Singapore's resident population aged 65 and above grew by 3.1% year-on-year, 
    continuing a structural trend.
  - Influenza activity was elevated in January and February 2026, with a 28% increase 
    in GP flu consultations compared to Q4 2025.

HOSPITAL OPERATIONS:
  - Average length of stay (ALOS) dropped from 4.3 days (Q4 2025) to 3.9 days (Q1 2026), 
    suggesting earlier discharges.
  - Bed occupancy rate reached 91% in January 2026, up from 84% in Q4 2025.
  - Three hospitals reported nursing staff shortages of 8–12% in January and February 
    due to post-holiday leave and sick leave spikes.

POLICY & CARE TRANSITION:
  - A new fast-track discharge protocol was piloted at two clusters in Q1 2026 to 
    relieve bed pressure. Post-discharge follow-up rates under this protocol are 
    not yet fully measured.
  - Community Health Assist Scheme (CHAS) GP clinic capacity was stretched in Q1 2026, 
    with average wait times for follow-up appointments increasing from 3 days to 7 days.

PATIENT FEEDBACK (social media and MOH feedback portal — 500 mentions analysed):
  - 34% of readmission-related feedback mentioned feeling 'discharged too early'.
  - 21% mentioned difficulty getting timely follow-up appointments with GPs or specialists.
  - 15% mentioned confusion about post-discharge medication instructions.

Based on these signals, what are the most likely root causes of the readmission spike?
"""

kw.listen_and_act(diagnostic_briefing, max_content_length=4000)
```

## Step 3 — Predictive Analysis

Ask KW to forecast whether the elevated readmission rate will persist into Q2 2026.

```python
predictive_briefing = """
Thank you. Looking ahead to Q2 2026:

  - The fast-track discharge protocol is scheduled to expand to all public hospital 
    clusters in April 2026 unless MOH intervenes.
  - CHAS GP capacity is expected to remain constrained through at least May 2026 
    due to an ongoing manpower gap.
  - Singapore's dengue season typically peaks in Q2, which may further stress 
    hospital capacity.
  - No new community step-down care capacity (community hospitals, home care) 
    is scheduled to come online before July 2026.

Based on these forward-looking signals and the diagnostic factors you identified, 
what is your prediction for the readmission trajectory in Q2 2026? 
What range would you expect and with what confidence?
"""

kw.listen_and_act(predictive_briefing, max_content_length=4000)
```

## Step 4 — Prescriptive Analysis

Ask KW to recommend concrete actions for MOH.

```python
prescriptive_briefing = """
Given everything you have analysed, what are your specific recommendations for MOH?

Please structure your recommendations by:
1. Immediate actions (within the next 30 days)
2. Medium-term actions (Q2 2026)
3. Longer-term structural recommendations

Also advise what data we should be collecting or monitoring going forward 
to detect early warning signals of readmission rate increases.
"""

kw.listen_and_act(prescriptive_briefing, max_content_length=4000)
```

## Extract Structured Findings

Use TinyTroupe's `ResultsExtractor` to pull a structured summary of KW's analysis.

```python
extractor = ResultsExtractor()

findings = extractor.extract_results_from_agent(
    kw,
    extraction_objective="""
        Extract KW's structured analytical findings in four sections:
        1. Descriptive findings: the confirmed statistics and significance test result
        2. Diagnostic findings: ranked root causes of the readmission spike
        3. Predictive findings: forecast for Q2 2026 readmission rate with confidence range
        4. Prescriptive recommendations: immediate, medium-term, and structural actions
    """,
    situation="KW has just completed a 4-tier analysis of Singapore hospital readmission data.",
    fields=["descriptive_findings", "diagnostic_findings", "predictive_findings", "prescriptive_recommendations"],
    verbose=True
)

print(json.dumps(findings, indent=2))
```
